# Resume-to-job ML prototype

Starter notebook for exploring skill overlap and ranking. Replace the baseline with calibrated evaluation before using real candidate data.

```python
import json
from pathlib import Path

data_dir = Path('data')
resumes = json.loads((data_dir / 'sample_resumes.json').read_text())
jobs = json.loads((data_dir / 'sample_jobs.json').read_text())
resumes, jobs
```

```python
def overlap_score(resume, job):
    resume_skills = {skill.casefold() for skill in resume['skills']}
    job_skills = {skill.casefold() for skill in job['skills']}
    return len(resume_skills & job_skills) / max(len(job_skills), 1)

[(resume['id'], job['id'], overlap_score(resume, job))
 for resume in resumes for job in jobs]
```
